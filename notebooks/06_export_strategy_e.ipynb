```python
# Cell 1: Load your trained Strategy E model
import tensorflow as tf
from pathlib import Path

ROOT_DIR = Path.cwd()
if not (ROOT_DIR / 'models').exists():
    ROOT_DIR = ROOT_DIR.parent

MODEL_PATH = ROOT_DIR / 'models' / 'model_E_selective_V3_final.h5'
MODEL_OUTPUT_PATH = ROOT_DIR / 'models' / 'strategy_e_best.h5'
MODEL_OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)

model_to_save = tf.keras.models.load_model(str(MODEL_PATH))
print(f"Loaded model from: {MODEL_PATH}")

model_to_save.save(str(MODEL_OUTPUT_PATH))
print(f"✓ Model saved to: {MODEL_OUTPUT_PATH}")

# Also save model summary
with open(ROOT_DIR / 'models' / 'strategy_e_architecture.txt', 'w') as f:
    model_to_save.summary(print_fn=lambda x: f.write(x + '\n'))

print(f"✓ Architecture saved to: {ROOT_DIR / 'models' / 'strategy_e_architecture.txt'}")

# Verify the model can be loaded
verify_model = tf.keras.models.load_model(str(MODEL_OUTPUT_PATH))
print(f"✓ Model verification successful!")
print(f"  Input shape: {verify_model.input_shape}")
print(f"  Output shape: {verify_model.output_shape}")
```
